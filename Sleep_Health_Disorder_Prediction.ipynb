#Converting categorical values to Numericals

Automated Conversion using Label Encoder

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df1['Sleep Disorder'] = le.fit_transform(df1['Sleep Disorder'])
df1.head()
```

Since the Column named Blood Pressure has two different numeric values separated by '/' lets convert it to two different column has systolic BP and Diastolic BP

#Selecting the X has dependent/independent columns and Y has target column

```python
# Define the target variable (y)
y = df1['Sleep Disorder']

# Define the features (x) by dropping the target and irrelevant columns
x = df1.drop('Sleep Disorder', axis=1)
#We'll drop 'Person ID' as it's just an identifier, and the original 'Blood Pressure' column because we've replaced it with 'Systolic BP' and 'Diastolic BP'

print("Shape of x (features):", x.shape)
print("Shape of y (target):", y.shape)

x.head()
y.head(20)
```

#Building an ANN Model

Since Neural networks couldnt able to interpret label encoders relationship by treating as an Ordinal relationship. So we use One hot Encoding which encodes in binary vector

Also there are more number of distinct outcomes are in target column so it is best adviced to use softmax function

```python
from tensorflow.keras.utils import to_categorical
y_train_one_hot = to_categorical(y_train, num_classes=len(y.unique()))
y_test_one_hot = to_categorical(y_test, num_classes=len(y.unique()))
```

```python
import tensorflow as tf
from tensorflow import keras

# Define a larger ANN architecture
model = keras.Sequential([
    keras.layers.Dense(128, activation='relu', input_shape=(x_train.shape[1],)), # Increased to 128 neurons
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(32, activation='relu'),
    keras.layers.Dense(16, activation='relu'),                                # Hidden layer
    keras.layers.Dense(len(y.unique()), activation='softmax')                # Output layer
])

# Compile the model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Train the model
history = model.fit(x_train, y_train_one_hot,
                    epochs=100,
                    validation_data=(x_test, y_test_one_hot),
                    verbose=1)
```

```python
# Evaluate the model on the test set
loss, accuracy = model.evaluate(x_test, y_test_one_hot)
print(f"\nANN Test Accuracy: {accuracy*100:.2f}%")
```
